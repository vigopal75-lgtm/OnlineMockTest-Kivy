# Online Mock Test — Kivy Android APK
এই Notebook Google Colab-এ চালিয়ে আপনার Kivy Mock Test App-এর APK তৈরি করুন। প্রথম Build-এ Android SDK/NDK ডাউনলোড হবে, তাই সময় লাগতে পারে।


```python
from google.colab import files
uploaded = files.upload()  # OnlineMockTest_Kivy_Colab_APK.zip নির্বাচন করুন
import zipfile, os
zip_name = next(name for name in uploaded if name.lower().endswith('.zip'))
with zipfile.ZipFile(zip_name) as z:
    z.extractall('/content')
os.chdir('/content/OnlineMockTest_Kivy')
print('Project files:', os.listdir())
```
